```python
def print_numbers_up_to_n(n: int) -> None:
    """Print all numbers from 1 to n.
    
    Args:
        n (int): The upper limit value (inclusive).
    """
    for i in range(1, n + 1):
        print(i)
```

```python

# Example usage of the function
print_numbers_up_to_n(10)
```
